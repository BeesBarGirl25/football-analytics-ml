```python
import matplotlib.pyplot as plt
from mplsoccer import Pitch
from matplotlib.patches import Rectangle
```

```python
from football_analytics.config import X_DEF, X_ATT, Y_LEFT, Y_CENTRE, Y_LEFT_HALF, Y_LEFT_WING, Y_RIGHT_HALF, Y_RIGHT_WING, X_DEEP_DEF, X_BUILD_UP, X_MIDFIELD, X_ADV_MID, X_FINAL_THIRD
```

```python
pitch = Pitch(pitch_type="statsbomb", line_zorder=2)
PITCH_L=120
PITCH_W=80
```

```python
def vline(x, label = None, y = 78):
    ax.plot([x,x], [0,80], zorder=3)
    if label:
        ax.text(x, y, label, ha="center", va="bottom", fontsize=9)

def hline(y, label = None, x = 118):
    ax.plot([0,120], [y,y], zorder=3)
    if label:
        ax.text(x, y, label, ha="right", va="center", fontsize=9)

def _add_zone(ax, x0, y0, w, h, label=None, facecolor=None, alpha=0.22, lw=1.6):
    rect = Rectangle(
        (x0, y0), w, h,
        linewidth=lw,
        edgecolor="black",
        facecolor=facecolor,
        alpha=alpha,
        zorder=1
    )
    ax.add_patch(rect)

    if label:
        ax.text(
            x0 + w / 2,
            y0 + h / 2,
            label,
            ha="center",
            va="center",
            fontsize=10,
            weight="bold",
            zorder=10
        )
```

```python
def plot_pitch_thirds(savepath=None, figsize=(11, 7), title="Pitch Thirds"):
    """Highlight defensive/midfield/attacking thirds."""
    pitch = Pitch(pitch_type="statsbomb", line_zorder=2)
    fig, ax = pitch.draw(figsize=figsize)

    thirds = [
        (0, 0, X_DEF, PITCH_W, "Defensive", "#b7d4ea"),
        (X_DEF, 0, X_ATT - X_DEF, PITCH_W, "Midfield", "#dbead5"),
        (X_ATT, 0, PITCH_L - X_ATT, PITCH_W, "Attacking", "#f4d6d6"),
    ]
    for x0, y0, w, h, lab, col in thirds:
        _add_zone(ax, x0, y0, w, h, label=lab, facecolor=col)

    ax.set_title(title)
    if savepath:
        fig.savefig(savepath, dpi=200, bbox_inches="tight")
        plt.close(fig)
    return fig, ax


def plot_lateral_channels(savepath=None, figsize=(11, 7), title="Lateral Channels"):
    """Highlight left/central/right channels."""
    pitch = Pitch(pitch_type="statsbomb", line_zorder=2)
    fig, ax = pitch.draw(figsize=figsize)

    channels = [
        (0, 0, PITCH_L, Y_LEFT, "Left", "#b7d4ea"),
        (0, Y_LEFT, PITCH_L, Y_CENTRE, "Central", "#dbead5"),
        (0, Y_CENTRE, PITCH_L, PITCH_W - Y_CENTRE, "Right", "#f4d6d6"),
    ]
    for x0, y0, w, h, lab, col in channels:
        _add_zone(ax, x0, y0, w, h, label=lab, facecolor=col)

    ax.set_title(title)
    if savepath:
        fig.savefig(savepath, dpi=200, bbox_inches="tight")
        plt.close(fig)
    return fig, ax

    
def plot_halfspace_lanes(savepath=None, figsize=(11, 7), title="Half-spaces and Lanes"):
    """Highlight wing/half-space/central/half-space (your y-band system)."""
    pitch = Pitch(pitch_type="statsbomb", line_zorder=2)
    fig, ax = pitch.draw(figsize=figsize)

    lanes = [
        (0, 0, PITCH_L, Y_LEFT_WING, "Left wing", "#b7d4ea"),
        (0, Y_LEFT_WING, PITCH_L, Y_LEFT_HALF - Y_LEFT_WING, "Left half-space", "#dbead5"),
        (0, Y_LEFT_HALF, PITCH_L, Y_RIGHT_HALF - Y_LEFT_HALF, "Right half-space", "#f4d6d6"),
        (0, Y_RIGHT_HALF, PITCH_L, Y_RIGHT_WING - Y_RIGHT_HALF, "Right Wing", "#dbead5"),
    ]
    for x0, y0, w, h, lab, col in lanes:
        _add_zone(ax, x0, y0, w, h, label=lab, facecolor=col)

    ax.set_title(title)
    if savepath:
        fig.savefig(savepath, dpi=200, bbox_inches="tight")
        plt.close(fig)
    return fig, ax

def plot_depth_bands(savepath=None, figsize=(11, 7), title="Depth Bands"):
    """Highlight your x-band phase-of-play system."""
    pitch = Pitch(pitch_type="statsbomb", line_zorder=2)
    fig, ax = pitch.draw(figsize=figsize)

    bands = [
        (0, 0, X_DEEP_DEF, PITCH_W, "Deep defence", "#b7d4ea"),
        (X_DEEP_DEF, 0, X_BUILD_UP - X_DEEP_DEF, PITCH_W, "Build-up", "#dbead5"),
        (X_BUILD_UP, 0, X_MIDFIELD - X_BUILD_UP, PITCH_W, "Midfield", "#f4d6d6"),
        (X_MIDFIELD, 0, X_ADV_MID - X_MIDFIELD, PITCH_W, "Advanced midfield", "#dbead5"),
        (X_ADV_MID, 0, X_FINAL_THIRD - X_ADV_MID, PITCH_W, "Final third", "#b7d4ea"),
    ]
    for x0, y0, w, h, lab, col in bands:
        _add_zone(ax, x0, y0, w, h, label=lab, facecolor=col)

    ax.set_title(title)
    if savepath:
        fig.savefig(savepath, dpi=200, bbox_inches="tight")
        plt.close(fig)
    return fig, ax

def plot_positional_grid_3x3(savepath=None, figsize=(11, 7), title="3×3 Positional Grid"):
    """Highlight thirds × channels and label dl/dc/.../ar."""
    pitch = Pitch(pitch_type="statsbomb", line_zorder=2)
    fig, ax = pitch.draw(figsize=figsize)

    x_edges = [0, X_DEF, X_ATT, PITCH_L]
    y_edges = [0, Y_LEFT, Y_CENTRE, PITCH_W]

    zone_names = [
        ["dl", "dc", "dr"],
        ["ml", "mc", "mr"],
        ["al", "ac", "ar"],
    ]

    # Row-based palette (def/mid/att) for visual grouping
    row_cols = ["#b7d4ea", "#dbead5", "#f4d6d6"]

    for i in range(3):
        for j in range(3):
            x0, x1 = x_edges[i], x_edges[i + 1]
            y0, y1 = y_edges[j], y_edges[j + 1]
            _add_zone(
                ax,
                x0, y0,
                x1 - x0,
                y1 - y0,
                label=zone_names[i][j],
                facecolor=row_cols[i],
            )

    ax.set_title(title)
    if savepath:
        fig.savefig(savepath, dpi=200, bbox_inches="tight")
        plt.close(fig)
    return fig, ax
```

```python
plot_pitch_thirds("01_thirds.png")
plot_lateral_channels("02_channels.png")
plot_halfspace_lanes("03_halfspaces.png")
plot_depth_bands("04_depth_bands.png")
plot_positional_grid_3x3("05_grid_3x3.png")

```
